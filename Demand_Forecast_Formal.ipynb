## Theoretical Background

### Definition of Demand Management
Demand management is the process of understanding, planning, and influencing 
customer demand for products or services. In a supply chain context, it involves 
collecting and analyzing historical demand data, identifying patterns, and using 
that information to make informed decisions about production, inventory, and 
distribution.

A key component of demand management is demand forecasting, which refers to 
the process of estimating future customer demand based on historical data and 
analytical methods. Accurate forecasting allows companies to align their supply 
with expected demand, reducing uncertainty and improving overall supply chain 
efficiency. In global supply chains where lead times are long, demand forecasting 
becomes especially critical as decisions must be committed to well before actual 
demand is realized.

### Impacts of Poor Demand Forecasting
When demand forecasting is inaccurate, it can have significant negative 
consequences across the entire supply chain. One of the most common outcomes 
is stockouts, which occur when inventory runs out before new stock arrives. 
This leads to unfulfilled customer orders, lost revenue, and damaged customer 
relationships.

On the opposite end, overestimating demand results in excess inventory 
accumulating in warehouses. This ties up capital that could be used elsewhere 
and increases holding and storage costs. Over time, excess stock may also 
become obsolete, resulting in further financial losses.

Poor forecasting also disrupts production planning. When demand signals are 
unreliable, manufacturing schedules become unstable, leading to inefficient 
use of resources, rushed production, or unnecessary downtime. In a global 
supply chain with long shipping times, these disruptions are even harder to 
recover from, as there is little time to react once a forecasting error 
is identified.

### Benefits of Accurate Forecasting
Accurate demand forecasting brings significant benefits to both operational 
and strategic aspects of a business. From an operational perspective, it allows 
warehouses to maintain optimal stock levels, ensuring that products are available 
when needed without holding unnecessary excess inventory. This leads to lower 
storage costs and more efficient use of warehouse space.

From a production standpoint, reliable forecasts enable more stable and 
consistent manufacturing schedules. This reduces waste, improves resource 
utilization, and minimizes the need for emergency adjustments. In a global 
supply chain, this stability is particularly valuable as it allows better 
coordination between manufacturing sites and distribution warehouses.

At a strategic level, accurate forecasting supports better decision-making 
across the organization. Management can plan procurement, logistics, and 
resource allocation with greater confidence. Ultimately, improved forecasting 
leads to higher service levels, stronger customer satisfaction, and a more 
resilient and responsive supply chain.

#### Clean Order_Demand (remove parentheses)

In standard accounting and financial reporting notation, values enclosed in parentheses are widely used to represent negative numbers. For example, (100) is interpreted as -100. This convention is recognised in accounting 
standards such as the International Financial Reporting Standards (IFRS) and is commonly used in financial statements, ledgers, and inventory records to indicate deductions, returns, or cancellations (Kieso et al., 2020).

Reference: Kieso, D. E., Weygandt, J. J., & Warfield, T. D. (2020). *Intermediate Accounting* (17th ed.). Wiley.

In the context of this dataset, the 5,899 rows containing parentheses in 
the Order_Demand column, such as (1), (24), and (100), are therefore 
interpreted as negative demand values. These entries likely represent 
order cancellations or product returns, where a customer placed an order 
but subsequently cancelled it before or after fulfillment.

#### What business insights from ABC product segmentation analysis?

Answer: Segment by demand frequency, size, and seasonality to create ABC categories: A (high value, seasonal), B (stable), C (low value). Focus inventory on A-class customers.

```python
product_demand = df_new.groupby('Product_Code')['Order_Demand'].sum().sort_values(ascending=False)

# Percentage contribution
cum_pct = product_demand.cumsum() / product_demand.sum()

abc = pd.DataFrame({
    'Demand': product_demand,
    'Cumulative %': cum_pct
})

# Assign ABC categories
abc['Category'] = pd.cut(abc['Cumulative %'],
                         bins=[0, 0.8, 0.95, 1],
                         labels=['A', 'B', 'C'])

# Plot
abc['Category'].value_counts().plot(kind='bar')
plt.title("ABC Analysis of Products")
plt.show()
```

The ABC analysis classifies all 2,160 products based on their contribution 
to total demand:

- **Class A** — Top products contributing to 80% of total demand. 
  These are critical products that require the highest forecasting accuracy 
  and safety stock levels.
- **Class B** — Products contributing to the next 15% of demand. 
  These require regular monitoring but less intensive planning.
- **Class C** — Remaining products contributing only 5% of demand. 
  These are low priority and can be managed with simpler replenishment rules.

This segmentation allows the company to focus forecasting efforts on Class A 
products rather than treating all 2,160 products equally.

### Stationarity Analysis — ADF Test


**What is Stationarity?**
A time series is considered **stationary** when its statistical properties — 
such as mean and variance — remain constant over time. Stationarity is a 
fundamental requirement for many time series forecasting models such as ARMA and ARIMA.

To formally test stationarity, we apply the **Augmented Dickey-Fuller (ADF) Test**:
- **Null Hypothesis (H0):** The time series is NOT stationary
- If **p-value ≤ 0.05** → Reject H0 → Series IS stationary
- If **p-value > 0.05** → Fail to reject H0 → Series is NOT stationary

```python
from statsmodels.tsa.stattools import adfuller

warehouses = ['Whse_J', 'Whse_S', 'Whse_C', 'Whse_A']
colors = {'Whse_J': 'green', 'Whse_S': 'red', 'Whse_C': 'orange', 'Whse_A': 'blue'}

fig, axes = plt.subplots(2, 2, figsize=(18, 10))
axes = axes.flatten()

for i, wh in enumerate(warehouses):
    df_wh = df_new[(df_new['Warehouse'] == wh) &
                   (df_new['Date'].dt.year >= 2012) &
                   (df_new['Date'].dt.year <= 2016)]
    
    monthly = df_wh.groupby(df_wh['Date'].dt.to_period('M'))['Order_Demand'].sum()
    
    # ADF Test
    result = adfuller(monthly)
    pval = result[1]
    status = 'Stationary' if pval <= 0.05 else 'Not Stationary'
    
    # Plot
    monthly.plot(ax=axes[i], color=colors[wh])
    axes[i].set_title(f'{wh} — Monthly Demand\n(ADF p-value: {pval:.4f} | {status})')
    axes[i].set_xlabel('Date')
    axes[i].set_ylabel('Total Demand')

plt.suptitle('Monthly Demand Over Time — All Warehouses', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

#### Results

| Warehouse | ADF p-value | Result |
|---|---|---|
| Whse_J | > 0.05 | Not Stationary |
| Whse_S | > 0.05 | Not Stationary |
| Whse_C | ≤ 0.05 | Stationary |
| Whse_A | ≤ 0.05 | Stationary |

throughout 2012–2016 with no significant upward or downward drift. 
This means their demand patterns are stable and predictable, 
and ARMA models can be applied directly for forecasting.

**Whse_J and Whse_S (Not Stationary)**
Whse_J shows a clear upward trend — demand in 2016 is noticeably

Whse_S shows the opposite — a clear downward trend from 2012 to 2016,

### Time Series Decomposition

Time series decomposition breaks down each warehouse's monthly demand into 
three components to better understand the underlying demand behavior:

- **Trend:** The long-term direction of demand (increasing, decreasing, or stable)
- **Seasonality:** Repeating patterns that occur at regular intervals (every 12 months)
- **Residual:** Random noise remaining after removing trend and seasonality

---

```python
from statsmodels.tsa.seasonal import seasonal_decompose
```

#### Whse_J (Green)

```python
df_wh = df_new[(df_new['Warehouse'] == 'Whse_J') &
               (df_new['Date'].dt.year >= 2012) &
               (df_new['Date'].dt.year <= 2016)]

monthly = df_wh.groupby(df_wh['Date'].dt.to_period('M'))['Order_Demand'].sum()
monthly.index = monthly.index.to_timestamp()

decomposition = seasonal_decompose(monthly, model='additive', period=12)

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

decomposition.observed.plot(ax=axes[0][0], color='green')
axes[0][0].set_title('Observed')
axes[0][0].set_ylabel('Demand')

decomposition.trend.plot(ax=axes[0][1], color='green')
axes[0][1].set_title('Trend')
axes[0][1].set_ylabel('Demand')

decomposition.seasonal.plot(ax=axes[1][0], color='green')
axes[1][0].set_title('Seasonality')
axes[1][0].set_ylabel('Demand')

decomposition.resid.plot(ax=axes[1][1], color='green')
axes[1][1].set_title('Residual')
axes[1][1].set_ylabel('Demand')

plt.suptitle('Whse_J — Time Series Decomposition', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Trend:** Whse_J displays a clear upward trend from approximately 5.1M units 
in 2012, peaking around 6M in 2015 before slightly declining toward 2016. 
This consistent upward drift confirms the ADF test result of non-stationarity, 
indicating that demand is systematically growing over time.

**Seasonality:** Strong and consistent seasonal patterns are observed throughout 
the entire period, with peaks and troughs occurring at the same time each year. 
This suggests predictable high and low demand periods that can be planned for.

**Residual:** Residuals remain small and close to zero, indicating that the 
trend and seasonal components effectively explain most of the demand variation 
in Whse_J.
