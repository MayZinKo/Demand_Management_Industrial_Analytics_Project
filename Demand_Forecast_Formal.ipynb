## Theoretical Background

### Definition of Demand Management
Demand management is the process of understanding, planning, and influencing 
customer demand for products or services. In a supply chain context, it involves 
collecting and analyzing historical demand data, identifying patterns, and using 
that information to make informed decisions about production, inventory, and 
distribution.

A key component of demand management is demand forecasting, which refers to 
the process of estimating future customer demand based on historical data and 
analytical methods. Accurate forecasting allows companies to align their supply 
with expected demand, reducing uncertainty and improving overall supply chain 
efficiency. In global supply chains where lead times are long, demand forecasting 
becomes especially critical as decisions must be committed to well before actual 
demand is realized.

### Impacts of Poor Demand Forecasting
When demand forecasting is inaccurate, it can have significant negative 
consequences across the entire supply chain. One of the most common outcomes 
is stockouts, which occur when inventory runs out before new stock arrives. 
This leads to unfulfilled customer orders, lost revenue, and damaged customer 
relationships.

On the opposite end, overestimating demand results in excess inventory 
accumulating in warehouses. This ties up capital that could be used elsewhere 
and increases holding and storage costs. Over time, excess stock may also 
become obsolete, resulting in further financial losses.

Poor forecasting also disrupts production planning. When demand signals are 
unreliable, manufacturing schedules become unstable, leading to inefficient 
use of resources, rushed production, or unnecessary downtime. In a global 
supply chain with long shipping times, these disruptions are even harder to 
recover from, as there is little time to react once a forecasting error 
is identified.

### Benefits of Accurate Forecasting
Accurate demand forecasting brings significant benefits to both operational 
and strategic aspects of a business. From an operational perspective, it allows 
warehouses to maintain optimal stock levels, ensuring that products are available 
when needed without holding unnecessary excess inventory. This leads to lower 
storage costs and more efficient use of warehouse space.

From a production standpoint, reliable forecasts enable more stable and 
consistent manufacturing schedules. This reduces waste, improves resource 
utilization, and minimizes the need for emergency adjustments. In a global 
supply chain, this stability is particularly valuable as it allows better 
coordination between manufacturing sites and distribution warehouses.

At a strategic level, accurate forecasting supports better decision-making 
across the organization. Management can plan procurement, logistics, and 
resource allocation with greater confidence. Ultimately, improved forecasting 
leads to higher service levels, stronger customer satisfaction, and a more 
resilient and responsive supply chain.

#### Clean Order_Demand (remove parentheses)

In standard accounting and financial reporting notation, values enclosed in parentheses are widely used to represent negative numbers. For example, (100) is interpreted as -100. This convention is recognised in accounting 
standards such as the International Financial Reporting Standards (IFRS) and is commonly used in financial statements, ledgers, and inventory records to indicate deductions, returns, or cancellations (Kieso et al., 2020).

Reference: Kieso, D. E., Weygandt, J. J., & Warfield, T. D. (2020). *Intermediate Accounting* (17th ed.). Wiley.

In the context of this dataset, the 5,899 rows containing parentheses in 
the Order_Demand column, such as (1), (24), and (100), are therefore 
interpreted as negative demand values. These entries likely represent 
order cancellations or product returns, where a customer placed an order 
but subsequently cancelled it before or after fulfillment. 

````python
df_cleaned = df_cleaned[~df_cleaned['Order_Demand'].str.contains('\(', regex=True)]
````

### ABC analysis

What ABC analysis means
ABC analysis is a Pareto-based inventory method:
A items = a small number of products that create most demand.
B items = medium importance.
C items = many low-demand products.
This is useful because inventory control should focus more attention on A items and less on C items


````python
abc = (
    df_new.groupby('Product_Code')['Order_Demand']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

abc['cum_demand_pct'] = abc['Order_Demand'].cumsum() / abc['Order_Demand'].sum() * 100

def classify_abc(x):
    if x <= 80:
        return 'A'
    elif x <= 95:
        return 'B'
    else:
        return 'C'

abc['ABC_Class'] = abc['cum_demand_pct'].apply(classify_abc)

print(abc.head(20))
print("\nABC class counts:")
print(abc['ABC_Class'].value_counts())
````

#### ABC analysis results
A products: top priority, should rarely stock out.
B products: moderate priority.
C products: low priority, avoid overstocking.

### Stationarity Analysis — ADF Test


**What is Stationarity?**
A time series is considered **stationary** when its statistical properties — 
such as mean and variance — remain constant over time. Stationarity is a 
fundamental requirement for many time series forecasting models such as ARMA and ARIMA.

To formally test stationarity, we apply the **Augmented Dickey-Fuller (ADF) Test**:
- **Null Hypothesis (H0):** The time series is NOT stationary
- If **p-value ≤ 0.05** → Reject H0 → Series IS stationary
- If **p-value > 0.05** → Fail to reject H0 → Series is NOT stationary

````python
from statsmodels.tsa.stattools import adfuller

warehouses = ['Whse_J', 'Whse_S', 'Whse_C', 'Whse_A']
colors = {'Whse_J': 'green', 'Whse_S': 'red', 'Whse_C': 'orange', 'Whse_A': 'blue'}

fig, axes = plt.subplots(2, 2, figsize=(18, 10))
axes = axes.flatten()

for i, wh in enumerate(warehouses):
    df_wh = df_new[(df_new['Warehouse'] == wh) &
                   (df_new['Date'].dt.year >= 2012) &
                   (df_new['Date'].dt.year <= 2016)]
    
    monthly = df_wh.groupby(df_wh['Date'].dt.to_period('M'))['Order_Demand'].sum()
    
    # ADF Test
    result = adfuller(monthly)
    pval = result[1]
    status = 'Stationary' if pval <= 0.05 else 'Not Stationary'
    
    # Plot
    monthly.plot(ax=axes[i], color=colors[wh])
    axes[i].set_title(f'{wh} — Monthly Demand\n(ADF p-value: {pval:.4f} | {status})')
    axes[i].set_xlabel('Date')
    axes[i].set_ylabel('Total Demand')

plt.suptitle('Monthly Demand Over Time — All Warehouses', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
````

#### Results

| Warehouse | ADF p-value | Result |
|---|---|---|
| Whse_J | > 0.05 | Not Stationary |
| Whse_S | > 0.05 | Not Stationary |
| Whse_C | ≤ 0.05 | Stationary |
| Whse_A | ≤ 0.05 | Stationary |

---

#### Interpretation

**Whse_C and Whse_A (Stationary)**
Both warehouses show demand fluctuating around a consistent mean level 
throughout 2012–2016 with no significant upward or downward drift. 
This means their demand patterns are stable and predictable, 
and ARMA models can be applied directly for forecasting.

**Whse_J and Whse_S (Not Stationary)**
Whse_J shows a clear upward trend — demand in 2016 is noticeably

Whse_S shows the opposite — a clear downward trend from 2012 to 2016, 
indicating declining regional demand. Since the mean is not constant, 
these series require differencing (d=1) before forecasting, 
making ARIMA the appropriate model rather than ARMA.



### Time Series Decomposition

Time series decomposition breaks down each warehouse's monthly demand into 
three components to better understand the underlying demand behavior:

- **Trend:** The long-term direction of demand (increasing, decreasing, or stable)
- **Seasonality:** Repeating patterns that occur at regular intervals (every 12 months)
- **Residual:** Random noise remaining after removing trend and seasonality

---

````python
from statsmodels.tsa.seasonal import seasonal_decompose
````

#### Whse_J (Green)

````python
df_wh = df_new[(df_new['Warehouse'] == 'Whse_J') &
               (df_new['Date'].dt.year >= 2012) &
               (df_new['Date'].dt.year <= 2016)]

monthly = df_wh.groupby(df_wh['Date'].dt.to_period('M'))['Order_Demand'].sum()
monthly.index = monthly.index.to_timestamp()

decomposition = seasonal_decompose(monthly, model='additive', period=12)

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

decomposition.observed.plot(ax=axes[0][0], color='green')
axes[0][0].set_title('Observed')
axes[0][0].set_ylabel('Demand')

decomposition.trend.plot(ax=axes[0][1], color='green')
axes[0][1].set_title('Trend')
axes[0][1].set_ylabel('Demand')

decomposition.seasonal.plot(ax=axes[1][0], color='green')
axes[1][0].set_title('Seasonality')
axes[1][0].set_ylabel('Demand')

decomposition.resid.plot(ax=axes[1][1], color='green')
axes[1][1].set_title('Residual')
axes[1][1].set_ylabel('Demand')

plt.suptitle('Whse_J — Time Series Decomposition', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

#### Model Selection

**Problem Type:** Time Series Forecasting

This project is a **time series forecasting** problem. The goal is to predict future product demand — a continuous numerical value — based on historical monthly demand data across four warehouses (Whse_A, Whse_C, Whse_J, Whse_S).

<br>

It falls under **supervised regression**, where the model learns from labelled historical data to predict future values. Unlike standard regression, the order of observations matters because demand at time *t* depends on its past values.

<br>

Different models handle this time dependency in different ways. Statistical models such as ARIMA and SARIMA are built specifically for time series and capture past value relationships internally. Machine learning models such as Random Forest and XGBoost are not natively time-aware, so past values must be manually provided as lag features (Lag_1, Lag_3, Lag_12) during feature engineering.

---

**Algorithms Considered**

The following algorithms were identified as suitable candidates for time series forecasting problems:

| Algorithm | Type | Brief Description |
|---|---|---|
| ARIMA / ARMA | Statistical | Models time dependency using past values and error terms. Requires stationarity check. |
| SARIMA | Statistical | Extension of ARIMA that handles seasonal patterns in the data. |
| Random Forest | Machine Learning | Ensemble of decision trees. Requires manual lag features to capture time dependency. |
| XGBoost | Machine Learning | Gradient boosting method. High accuracy but requires careful feature engineering. |
| Prophet | Machine Learning | Developed by Facebook. Handles trend and seasonality automatically. Easy to use. |
| LSTM | Deep Learning | Neural network designed for sequential data. Powerful but requires large data and complex setup. |
| ETS (Exponential Smoothing) | Statistical | Weights recent observations more heavily. Simple and effective for stable demand patterns. |

---

**Algorithm Comparison**

The following table compares candidate algorithms across five criteria relevant to this project:

| Algorithm | Interpretability | Accuracy (this project) | Complexity | Data Size Required | Business Fit |
|---|---|---|---|---|---|
| ARIMA / ARMA | ✅ High — clear parameters (p,d,q), easy to explain | ✅ Good for stationary/non-stationary series | 🟡 Medium — requires stationarity check | ✅ Works well with small data (60 points) | ✅ Standard forecasting tool in industry |
| SARIMA | ✅ High — extension of ARIMA with seasonal term | ✅ Good when seasonality is present | 🟡 Medium — adds seasonal parameters | ✅ Works well with small data | ✅ Suitable when demand has seasonal patterns |
| Random Forest | 🟡 Medium — feature importance available but not fully transparent | ✅ Good with engineered lag features | 🟡 Medium — requires manual feature engineering | 🟡 Moderate — needs enough rows for tree splitting | ✅ Flexible, handles non-linear demand patterns |
| XGBoost | 🟡 Medium — SHAP values help but still complex | ✅ High accuracy on tabular data | 🔴 High — hyperparameter tuning required | 🟡 Moderate — benefits from larger datasets | 🟡 Powerful but harder to justify for small datasets |

**Key insight:** Given only 60 monthly data points per warehouse, models requiring large datasets (LSTM, XGBoost) are less suitable. ARIMA/ARMA and Random Forest offer a good balance of accuracy, interpretability, and suitability for small time series data.

#### Evaluation Metrics

Three standard forecasting evaluation metrics are used to assess and compare model performance on the 2016 holdout test set:

##### Mean Absolute Error (MAE)
$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

- Measures the **average absolute difference** between actual and predicted demand in the original units (number of orders)
- Easy to interpret: an MAE of 50,000 means the forecast is off by 50,000 units on average per month
- **Less sensitive to outliers** than RMSE
- Used to understand the typical forecasting error in business terms

##### Root Mean Squared Error (RMSE)
$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

- Measures the **square root of average squared errors**, expressed in the same units as demand
- **Penalizes large errors more heavily** than MAE — useful for catching months where the forecast was badly wrong
- Important for inventory planning where large under-forecasts cause serious stockouts

##### Mean Absolute Percentage Error (MAPE)
$$MAPE = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100$$

- Expresses error as a **percentage of actual demand** — scale-independent
- Allows **direct comparison across warehouses** of different sizes (e.g., comparing Whse_J with ~6M units vs Whse_A with ~0.4M units)
- Industry standard KPI for forecasting accuracy: MAPE < 10% is generally considered excellent for supply chain forecasting

| Metric | Unit | Strength | Use case |
|---|---|---|---|
| MAE | Orders / month | Interpretable, robust | Business reporting |
| RMSE | Orders / month | Penalizes large errors | Risk management |
| MAPE | % | Scale-free comparison | Cross-warehouse benchmarking |

### Model 1: ARIMA / ARMA (May Zin Ko)

#### What is ARIMA?

**ARIMA** stands for **AutoRegressive Integrated Moving Average**. It is one of the most widely used classical methods for time series forecasting.

It has three components controlled by parameters **(p, d, q)**:

| Parameter | Name | Role |
|---|---|---|
| **p** | AutoRegressive order | How many past values the model uses to predict the next value |
| **d** | Differencing order | How many times the series is differenced to achieve stationarity |
| **q** | Moving Average order | How many past forecast errors are used to correct the prediction |

When **d = 0**, the model reduces to **ARMA** — applied to already-stationary series (Whse_C and Whse_A).

**Why ARIMA for this dataset?**
The ADF test confirmed non-stationarity in Whse_J and Whse_S (requiring d=1), while time series decomposition revealed repeating seasonal structure. ARIMA is designed precisely for data with these characteristics.

#### Monthly Aggregation & Warehouse Configuration

Each warehouse's daily order records are aggregated into monthly total demand — the temporal resolution used for forecasting. The `d` parameter per warehouse is set based on the ADF stationarity results.

````python
# Warehouse configuration — d values from ADF test results (Task 5)
WAREHOUSE_CONFIG = {
    'Whse_J': {'d': 1, 'color': 'green',  'model_name': 'ARIMA'},  # non-stationary
    'Whse_S': {'d': 1, 'color': 'red',    'model_name': 'ARIMA'},  # non-stationary
    'Whse_C': {'d': 0, 'color': 'orange', 'model_name': 'ARMA'},   # stationary
    'Whse_A': {'d': 0, 'color': 'blue',   'model_name': 'ARMA'},   # stationary
}

# Aggregate daily demand to monthly per warehouse
warehouse_ts = {}
for wh in WAREHOUSE_CONFIG:
    df_wh = df_new[df_new['Warehouse'] == wh]
    monthly = (df_wh
               .groupby(df_wh['Date'].dt.to_period('M'))['Order_Demand']
               .sum())
    monthly.index = monthly.index.to_timestamp()
    warehouse_ts[wh] = monthly
    print(f"  {wh}: {len(monthly)} monthly observations")

print("\nTrain: 2012–2015 (48 months) | Test (holdout): 2016 (12 months)")
````

#### Automatic Parameter Selection — auto_arima

Rather than manually testing every combination of (p, d, q), `auto_arima` from the `pmdarima` library automatically searches for the optimal parameters using the **AIC (Akaike Information Criterion)** — a metric that balances model fit against complexity to prevent overfitting.

The `d` parameter is fixed per warehouse based on our ADF test results. Only `p` and `q` are searched in the range 0–4.

````python
from pmdarima import auto_arima

results = {}

for wh, cfg in WAREHOUSE_CONFIG.items():
    ts    = warehouse_ts[wh]
    train = ts[ts.index.year <= 2015]
    test  = ts[ts.index.year == 2016]

    print(f"\n--- {wh} ({cfg['model_name']}) ---")

    best = auto_arima(
        train,
        d=cfg['d'],           # fixed from ADF test results
        start_p=0, max_p=4,
        start_q=0, max_q=4,
        information_criterion='aic',   # minimize AIC
        seasonal=False,
        stepwise=True,                 # efficient stepwise search
        suppress_warnings=True,
        trace=True                     # shows models tried and AIC scores
    )

    print(f"Best order: {cfg['model_name']}{best.order}")
    print(f"AIC = {best.aic():.2f}")

    results[wh] = {
        'train':      train,
        'test':       test,
        'order':      best.order,
        'color':      cfg['color'],
        'model_name': cfg['model_name'],
    }
````

#### Model Training & 12-Month Forecast

Using the optimal (p, d, q) orders found by `auto_arima`, each warehouse model is fitted with `statsmodels ARIMA` and used to generate a 12-step ahead forecast covering all of 2016. A 95% confidence interval is also computed to show the uncertainty range of each forecast.

````python
from statsmodels.tsa.arima.model import ARIMA

for wh, r in results.items():

    # --- Fit the model ---
    fit = ARIMA(r['train'], order=r['order']).fit()
    results[wh]['fit'] = fit

    # --- Forecast 12 months ahead (2016) ---
    fc      = fit.get_forecast(steps=12)
    fc_mean = fc.predicted_mean
    fc_ci   = fc.conf_int(alpha=0.05)   # 95% confidence interval

    # Align forecast index to real 2016 calendar dates
    fc_mean.index = r['test'].index
    fc_ci.index   = r['test'].index

    results[wh]['fc_mean'] = fc_mean
    results[wh]['fc_ci']   = fc_ci

    # --- Compute evaluation metrics ---
    mae  = np.mean(np.abs(r['test'] - fc_mean))
    rmse = np.sqrt(np.mean((r['test'] - fc_mean) ** 2))
    mape = np.mean(np.abs((r['test'] - fc_mean) / r['test'])) * 100

    results[wh]['mae']  = mae
    results[wh]['rmse'] = rmse
    results[wh]['mape'] = mape

    print(f"{wh} {r['model_name']}{r['order']}  →  MAPE={mape:.1f}%  MAE={mae:,.0f}  RMSE={rmse:,.0f}")
````

#### ARIMA Forecast Visualization

Each subplot shows the full training history (2012–2015), the actual 2016 demand, the model forecast, and the 95% confidence interval. The vertical dotted line marks the train/test boundary.

````python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('ARMA / ARIMA Forecast vs Actual by Warehouse (2016)',
             fontsize=14, fontweight='bold')
axes = axes.flatten()

for i, (wh, r) in enumerate(results.items()):
    ax = axes[i]

    # Training history
    ax.plot(r['train'].index, r['train'].values,
            color=r['color'], linewidth=1.8, label='Train (2012–2015)')

    # Actual 2016 demand
    ax.plot(r['test'].index, r['test'].values,
            color='black', linewidth=2, marker='o', markersize=5,
            label='Actual 2016')

    # Forecast line
    ax.plot(r['fc_mean'].index, r['fc_mean'].values,
            color='crimson', linewidth=2, linestyle='--',
            marker='x', markersize=5, label='Forecast')

    # 95% confidence interval shading
    ax.fill_between(r['fc_ci'].index,
                    r['fc_ci'].iloc[:, 0],
                    r['fc_ci'].iloc[:, 1],
                    color='crimson', alpha=0.15, label='95% CI')

    # Train/test boundary
    ax.axvline(x=pd.Timestamp('2016-01-01'),
               color='gray', linestyle=':', linewidth=1.5)

    ax.set_title(f"{wh} — {r['model_name']}{r['order']}   MAPE={r['mape']:.1f}%",
                 fontweight='bold')
    ax.set_ylabel('Monthly Demand')
    ax.legend(fontsize=8)
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x/1e6:.1f}M'))
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

### Model 2: Random Forest Regressor (May Zin Ko)

#### What is Random Forest?

**Random Forest** is an ensemble machine learning method built from many **decision trees**. Each tree is trained on a random subset of the data and features, and the final prediction is the average across all trees.

Key hyperparameters used:

| Parameter | Value | Meaning |
|---|---|---|
| `n_estimators` | 100 | Number of decision trees in the forest |
| `random_state` | 42 | Seed for reproducibility |

**Why Random Forest for this dataset?**
Unlike ARIMA which models the time-series structure directly, Random Forest treats forecasting as a **supervised regression problem**. Past demand values (lags) and time-based features are used as input variables to predict future demand. This approach can capture non-linear relationships that ARIMA may miss, and handles all four warehouses simultaneously using warehouse as an encoded feature.

#### Feature Engineering

Random Forest requires explicit features — the model cannot learn from time-ordering alone. The following features are engineered from the monthly demand series:

| Feature | Description | Captures |
|---|---|---|
| `month` | Calendar month (1–12) | Seasonality |
| `year` | Calendar year (2012–2016) | Long-term trend |
| `lag_1` | Demand 1 month ago | Short-term momentum |
| `lag_3` | Demand 3 months ago | Quarterly pattern |
| `lag_12` | Demand 12 months ago | Year-over-year seasonality |
| `rolling_3` | 3-month rolling average | Short-term smoothed trend |
| `rolling_6` | 6-month rolling average | Medium-term smoothed trend |
| `warehouse_code` | Warehouse encoded as integer | Cross-warehouse differences |

All lag and rolling features are shifted to avoid data leakage — only information available *before* the forecast month is used.

````python
# Build feature table for all warehouses combined
all_data = []

for wh, ts in warehouse_ts.items():
    df_wh = ts.reset_index()
    df_wh.columns = ['Date', 'Demand']
    df_wh['Warehouse'] = wh

    # Time features
    df_wh['month'] = df_wh['Date'].dt.month
    df_wh['year']  = df_wh['Date'].dt.year

    # Lag features — past demand values (shifted to prevent leakage)
    df_wh['lag_1']  = df_wh['Demand'].shift(1)
    df_wh['lag_3']  = df_wh['Demand'].shift(3)
    df_wh['lag_12'] = df_wh['Demand'].shift(12)

    # Rolling averages — computed on already-shifted values
    df_wh['rolling_3'] = df_wh['Demand'].shift(1).rolling(3).mean()
    df_wh['rolling_6'] = df_wh['Demand'].shift(1).rolling(6).mean()

    all_data.append(df_wh)

# Combine all warehouses into one flat table
df_ml = pd.concat(all_data).dropna().reset_index(drop=True)

# Encode warehouse as integer (required by sklearn)
df_ml['warehouse_code'] = df_ml['Warehouse'].map(
    {'Whse_J': 0, 'Whse_S': 1, 'Whse_C': 2, 'Whse_A': 3}
)

print(f"Feature table shape: {df_ml.shape}")
print(f"\nSample rows:")
print(df_ml.head())
````

#### Train / Test Split

The same holdout strategy as ARIMA is applied — training on 2012–2015, testing on 2016 — ensuring a fair comparison between both models on identical data.

````python
from sklearn.ensemble import RandomForestRegressor

# Feature list
features = ['month', 'year', 'lag_1', 'lag_3', 'lag_12',
            'rolling_3', 'rolling_6', 'warehouse_code']

# Holdout split — identical to ARIMA
train_df = df_ml[df_ml['year'] <= 2015]
test_df  = df_ml[df_ml['year'] == 2016]

X_train = train_df[features]
y_train = train_df['Demand']

X_test  = test_df[features]
y_test  = test_df['Demand']

print(f"Train size : {len(X_train)} rows  ({len(X_train)//4} months × 4 warehouses)")
print(f"Test size  : {len(X_test)} rows  ({len(X_test)//4} months × 4 warehouses)")
````

#### Model Training & Evaluation

The Random Forest is trained on the full training set and evaluated per warehouse on the 2016 holdout data. Inputs are converted to NumPy arrays for compatibility with scikit-learn.

````python
# Convert to numpy for sklearn compatibility
X_train_np = np.array(X_train, dtype=np.float64)
y_train_np = np.array(y_train, dtype=np.float64)
X_test_np  = np.array(X_test,  dtype=np.float64)

# Train Random Forest
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train_np, y_train_np)
print("Model trained successfully!")

# Predict on 2016 holdout
y_pred = rf_model.predict(X_test_np)

# Store predictions in test dataframe
test_df = test_df.copy()
test_df['Predicted'] = y_pred

# Evaluate per warehouse
print(f"\n{'Warehouse':<12} {'MAE':>12} {'RMSE':>12} {'MAPE':>8}")
print("-" * 48)

for wh in ['Whse_J', 'Whse_S', 'Whse_C', 'Whse_A']:
    wh_df     = test_df[test_df['Warehouse'] == wh]
    actual    = wh_df['Demand']
    predicted = wh_df['Predicted']

    mae  = np.mean(np.abs(actual - predicted))
    rmse = np.sqrt(np.mean((actual - predicted) ** 2))
    mape = np.mean(np.abs((actual - predicted) / actual)) * 100

    print(f"{wh:<12} {mae:>12,.0f} {rmse:>12,.0f} {mape:>7.1f}%")
````

### Model 4: PROPHET MODEL

**What is Prophet?**

Prophet is a forecasting model designed for business time series with trend and seasonality. It is fast, automated, and especially convenient when forecasting needs to be repeated across several series with minimal manual tuning.

**Why Prophet?**

Prophet was chosen because it offers a strong business-oriented baseline for monthly demand forecasting. It is useful when the goal is to compare demand patterns across warehouses and understand which series are predictable in a practical planning sense

#### Prophet model on warehouse J, S, A, C 

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from prophet import Prophet
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = df_new.copy()
df['Date'] = pd.to_datetime(df['Date'])

results = []

for wh in df['Warehouse'].dropna().unique():
    wh_df = df[df['Warehouse'] == wh].copy()

    monthly = (
        wh_df.groupby(pd.Grouper(key='Date', freq='MS'))['Order_Demand']
        .sum()
        .reset_index()
        .sort_values('Date')
    )

    if len(monthly) < 12:
        print(f"Skip {wh}: not enough data")
        continue

    prophet_df = monthly.rename(columns={'Date': 'ds', 'Order_Demand': 'y'})
    train = prophet_df[prophet_df['ds'].dt.year <= 2015].copy()
    test = prophet_df[prophet_df['ds'].dt.year == 2016].copy()

    if len(train) == 0 or len(test) == 0:
        print(f"Skip {wh}: no train/test split")
        continue

    model = Prophet(yearly_seasonality=True, weekly_seasonality=False, daily_seasonality=False)
    model.fit(train)

    future = model.make_future_dataframe(periods=len(test), freq='MS', include_history=False)
    forecast = model.predict(future)

    pred = forecast['yhat'].values[:len(test)]
    y_true = test['y'].values

    mae = mean_absolute_error(y_true, pred)
    rmse = np.sqrt(mean_squared_error(y_true, pred))
    mape = np.mean(np.abs((y_true - pred) / np.where(y_true == 0, np.nan, y_true))) * 100

    print(f"\nWarehouse: {wh}")
    print("Prophet MAE:", mae)
    print("Prophet RMSE:", rmse)
    print("Prophet MAPE:", mape)

    plt.figure(figsize=(12,4))
    plt.plot(train['ds'], train['y'], label='Train')
    plt.plot(test['ds'], test['y'], label='Actual 2016')
    plt.plot(test['ds'], pred, label='Prophet Forecast')
    plt.axvline(pd.Timestamp('2016-01-01'), color='red', linestyle='--')
    plt.title(f'{wh} - Prophet Forecast')
    plt.legend()
    plt.tight_layout()
    plt.show()

    results.append([wh, mae, rmse, mape])

results_df = pd.DataFrame(results, columns=['Warehouse', 'MAE', 'RMSE', 'MAPE'])
print(results_df)
````

Prophet showed the strongest forecasting performance for Warehouse J and Warehouse C, indicating stable demand patterns that support inventory planning and replenishment decisions. In contrast, Warehouse S showed weak predictive performance, suggesting that demand is highly irregular and requires caution in operational use. Overall, the results demonstrate that forecasting quality differs across warehouses, so inventory strategies should be warehouse-specific rather than uniform.
