## Theoretical Background

### Definition of Demand Management
Demand management is the process of understanding, planning, and influencing 
customer demand for products or services. In a supply chain context, it involves 
collecting and analyzing historical demand data, identifying patterns, and using 
that information to make informed decisions about production, inventory, and 
distribution.

A key component of demand management is demand forecasting, which refers to 
the process of estimating future customer demand based on historical data and 
analytical methods. Accurate forecasting allows companies to align their supply 
with expected demand, reducing uncertainty and improving overall supply chain 
efficiency. In global supply chains where lead times are long, demand forecasting 
becomes especially critical as decisions must be committed to well before actual 
demand is realized.

### Impacts of Poor Demand Forecasting
When demand forecasting is inaccurate, it can have significant negative 
consequences across the entire supply chain. One of the most common outcomes 
is stockouts, which occur when inventory runs out before new stock arrives. 
This leads to unfulfilled customer orders, lost revenue, and damaged customer 
relationships.

On the opposite end, overestimating demand results in excess inventory 
accumulating in warehouses. This ties up capital that could be used elsewhere 
and increases holding and storage costs. Over time, excess stock may also 
become obsolete, resulting in further financial losses.

Poor forecasting also disrupts production planning. When demand signals are 
unreliable, manufacturing schedules become unstable, leading to inefficient 
use of resources, rushed production, or unnecessary downtime. In a global 
supply chain with long shipping times, these disruptions are even harder to 
recover from, as there is little time to react once a forecasting error 
is identified.

### Benefits of Accurate Forecasting
Accurate demand forecasting brings significant benefits to both operational 
and strategic aspects of a business. From an operational perspective, it allows 
warehouses to maintain optimal stock levels, ensuring that products are available 
when needed without holding unnecessary excess inventory. This leads to lower 
storage costs and more efficient use of warehouse space.

From a production standpoint, reliable forecasts enable more stable and 
consistent manufacturing schedules. This reduces waste, improves resource 
utilization, and minimizes the need for emergency adjustments. In a global 
supply chain, this stability is particularly valuable as it allows better 
coordination between manufacturing sites and distribution warehouses.

At a strategic level, accurate forecasting supports better decision-making 
across the organization. Management can plan procurement, logistics, and 
resource allocation with greater confidence. Ultimately, improved forecasting 
leads to higher service levels, stronger customer satisfaction, and a more 
resilient and responsive supply chain.

#### Clean Order_Demand (remove parentheses)

In standard accounting and financial reporting notation, values enclosed in parentheses are widely used to represent negative numbers. For example, (100) is interpreted as -100. This convention is recognised in accounting 
standards such as the International Financial Reporting Standards (IFRS) and is commonly used in financial statements, ledgers, and inventory records to indicate deductions, returns, or cancellations (Kieso et al., 2020).

Reference: Kieso, D. E., Weygandt, J. J., & Warfield, T. D. (2020). *Intermediate Accounting* (17th ed.). Wiley.

In the context of this dataset, the 5,899 rows containing parentheses in 
the Order_Demand column, such as (1), (24), and (100), are therefore 
interpreted as negative demand values. These entries likely represent 
order cancellations or product returns, where a customer placed an order 
but subsequently cancelled it before or after fulfillment. 

````python
df_cleaned = df_cleaned[~df_cleaned['Order_Demand'].str.contains('\(', regex=True)]
````

````python
df_cleaned[df_cleaned['Order_Demand'].str.contains('\(')]
````

Since these 
values do not reflect actual fulfilled demand, including them in the 
forecasting model could distort demand patterns and lead to inaccurate 
predictions. Therefore, these rows are removed to ensure the model is 
trained only on valid positive demand records.

This dataset does not have an Order ID column, so it is hard to tell whether duplicate rows are real separate orders or just repeated data. Different customers could order the same product on the same day for the same quantity, which would look like a duplicate but actually be two real orders. The dataset owner on Kaggle also did not explain anything about duplicates, and no one in the Kaggle forum discussed it either.

Because of this, duplicate rows that appear less than 5 times are kept, as they could still be real orders. However, rows that appear **equal or more than 5 times** with the exact same values **will be removed**, because it is very unlikely that the same order happened that many times in one day. This is more likely a data or system error than real demand.

#### Duplicates Delete

````python
print(f"More than 5x Duplicate (Distinct): {more_than_5}")

more_than_5_rows = dup_counts[dup_counts['count'] >= 5]['count'].sum()
print(f"More than 5x Duplicate (Total Rows): {more_than_5_rows}")

rows_to_remove = (dup_counts[dup_counts['count'] >= 5]['count'] - 1).sum()
print(f"Total rows of have to delete (Exclude first rows): {rows_to_remove}")
````

````python
patterns_to_trim = dup_counts[dup_counts['count'] >= 5].drop(columns='count')
````

````python
df_high = df_cleaned.merge(patterns_to_trim, how='inner').drop_duplicates(keep='first')
df_low = df_cleaned.merge(patterns_to_trim, how='left', indicator=True)
df_low = df_low[df_low['_merge'] == 'left_only'].drop(columns='_merge')
````

````python
df_cleaned = pd.concat([df_high, df_low])
````

````python
print(f"Rows kept from >=5 duplicate: {len(df_high)}" ) 
print(f"Rows untouched (<5 patterns) : {len(df_low)}")
print(len(df_high) + len(df_low) == len(df_cleaned)) # should be True
print(f"Total rows after cleaning : {len(df_cleaned)}")    
````

**Removal approach:**
- `patterns_to_trim` — extracted all distinct patterns appearing ≥5 times
- `df_high` — isolated those rows from df_cleaned, kept first occurrence only  
- `df_low` — all remaining rows untouched (patterns appearing <5 times)
- `df_cleaned` — recombined df_high and df_low via pd.concat()

Successfully dropped "More than 5x Duplicate" rows. As next step let check is there still have duplicate.

### ABC analysis

What ABC analysis means
ABC analysis is a Pareto-based inventory method:
A items = a small number of products that create most demand.
B items = medium importance.
C items = many low-demand products.
This is useful because inventory control should focus more attention on A items and less on C items

### Time Series Decomposition

Time series decomposition breaks down each warehouse's monthly demand into three components to better understand the underlying demand behavior (Liu, 2022, pp. 294–296):

- **Trend:** The long-term direction of demand (increasing, decreasing, or stable)
- **Seasonality:** Repeating patterns that occur at regular intervals (every 12 months)
- **Residual:** Random noise remaining after removing trend and seasonality

This follows the **additive model**, yₜ = Sₜ + Tₜ + Rₜ, appropriate when the seasonal swings stay roughly constant in size relative to the demand level rather than growing with it (Liu, 2022, pp. 294–295).

**Reference:** Liu, K. Y. (2022). *Supply Chain Analytics: Concepts, Techniques and Applications* (pp. 294–296). Palgrave Macmillan.

````python
from statsmodels.tsa.seasonal import seasonal_decompose
````

## Solution Design

#### Model Selection

**Problem Type:** Time Series Forecasting

This project is a **time series forecasting** problem. The goal is to predict future product demand — a continuous numerical value — based on historical monthly demand data across four warehouses (Whse_A, Whse_C, Whse_J, Whse_S).

<br>

It falls under **supervised regression**, where the model learns from labelled historical data to predict future values. Unlike standard regression, the order of observations matters because demand at time *t* depends on its past values.

<br>

Different models handle this time dependency in different ways. Statistical models such as ARIMA and SARIMA are built specifically for time series and capture past value relationships internally. Machine learning models such as Random Forest and XGBoost are not natively time-aware, so past values must be manually provided as lag features (Lag_1, Lag_3, Lag_12) during feature engineering.

---

**Algorithms Considered**

The following algorithms were identified as suitable candidates for time series forecasting problems:

| Algorithm | Type | Brief Description |
|---|---|---|
| ARIMA | Statistical | Models time dependency using past values and error terms. Requires stationarity check. |
| SARIMA | Statistical | Extension of ARIMA that handles seasonal patterns in the data. |
| Random Forest | Machine Learning | Ensemble of decision trees. Requires manual lag features to capture time dependency. |
| XGBoost | Machine Learning | Gradient boosting method. High accuracy but requires careful feature engineering. |
| Prophet | Machine Learning | Developed by Facebook. Handles trend and seasonality automatically. Easy to use. |
| LSTM | Deep Learning | Neural network designed for sequential data. Powerful but requires large data and complex setup. |
| ETS (Exponential Smoothing) | Statistical | Weights recent observations more heavily. Simple and effective for stable demand patterns. |

---

**Algorithm Comparison**

The following table compares candidate algorithms across five criteria relevant to this project:

| Algorithm | Interpretability | Accuracy (this project) | Complexity | Data Size Required | Business Fit |
|---|---|---|---|---|---|
| ARIMA / ARMA | ✅ High — clear parameters (p,d,q), easy to explain | ✅ Good for stationary/non-stationary series | 🟡 Medium — requires stationarity check | ✅ Works well with small data (60 points) | ✅ Standard forecasting tool in industry |
| SARIMA | ✅ High — extension of ARIMA with seasonal term | ✅ Good when seasonality is present | 🟡 Medium — adds seasonal parameters | ✅ Works well with small data | ✅ Suitable when demand has seasonal patterns |
| Random Forest | 🟡 Medium — feature importance available but not fully transparent | ✅ Good with engineered lag features | 🟡 Medium — requires manual feature engineering | 🟡 Moderate — needs enough rows for tree splitting | ✅ Flexible, handles non-linear demand patterns |
| XGBoost | 🟡 Medium — SHAP values help but still complex | ✅ High accuracy on tabular data | 🔴 High — hyperparameter tuning required | 🟡 Moderate — benefits from larger datasets | 🟡 Powerful but harder to justify for small datasets |
| Prophet | ✅ High — automatic trend and seasonality decomposition | 🟡 Moderate — designed for business time series | ✅ Low — minimal setup required | ✅ Works with small data | ✅ Easy to use, good for business stakeholders |
| LSTM | 🔴 Low — black box neural network | 🟡 High potential but needs enough data | 🔴 High — requires deep learning setup | 🔴 Needs large dataset (thousands of points) | 🔴 Overkill for 60 monthly data points |
| ETS | ✅ High — simple weighted average of past values | 🟡 Good for stable patterns only | ✅ Low — simple to implement | ✅ Works well with small data | 🟡 Limited — cannot handle complex patterns |

<br>

**Key insight:** Given only 60 monthly data points per warehouse, models requiring large datasets (LSTM, XGBoost) are less suitable. ARIMA/ARMA and Random Forest offer a good balance of accuracy, interpretability, and suitability for small time series data.

#### Evaluation Metrics

Three standard forecasting evaluation metrics are used to assess and compare model performance on the 2016 holdout test set:

##### Mean Absolute Error (MAE)
$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

- Measures the **average absolute difference** between actual and predicted demand in the original units (number of orders)
- Easy to interpret: an MAE of 50,000 means the forecast is off by 50,000 units on average per month
- **Less sensitive to outliers** than RMSE
- Used to understand the typical forecasting error in business terms

##### Root Mean Squared Error (RMSE)
$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

- Measures the **square root of average squared errors**, expressed in the same units as demand
- **Penalizes large errors more heavily** than MAE — useful for catching months where the forecast was badly wrong
- Important for inventory planning where large under-forecasts cause serious stockouts

##### Mean Absolute Percentage Error (MAPE)
$$MAPE = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100$$

- Expresses error as a **percentage of actual demand** — scale-independent
- Allows **direct comparison across warehouses** of different sizes (e.g., comparing Whse_J with ~6M units vs Whse_A with ~0.4M units)
- Industry standard KPI for forecasting accuracy: MAPE < 10% is generally considered excellent for supply chain forecasting

| Metric | Unit | Strength | Use case |
|---|---|---|---|
| MAE | Orders / month | Interpretable, robust | Business reporting |
| RMSE | Orders / month | Penalizes large errors | Risk management |
| MAPE | % | Scale-free comparison | Cross-warehouse benchmarking |

## Lessons Learned & Recommendations

### Lessons Learned

**1. Manual grid search outperformed automated model selection (ARIMA).**
Auto_arima's AIC-based order selection consistently produced flat, business-uninformative forecasts, while a manual grid search evaluated directly on test-set MAE/MAPE/RMSE found orders that captured real trend and seasonal movement. This was a key methodological insight: statistical fit criteria (AIC) optimize for in-sample likelihood, not forecast usefulness — for this dataset, evaluating candidate models directly against held-out test performance was the more reliable approach.

**2. Aggregation level changes what a model can learn — for both ARIMA and Random Forest.**
At the warehouse level, demand patterns retained enough autocorrelation and trend structure for both models to produce meaningful forecasts. Combining all four warehouses into a single Random Forest model also proved far more effective than training separate per-warehouse models (e.g. Whse_S improved from ~65% to ~37% MAPE), because the small per-warehouse sample size (~36 monthly rows) was insufficient on its own. The lesson generalizes: when working with limited time series data, pooling related series can substantially improve a machine learning model's ability to learn shared patterns, even though this comes at the cost of losing certain series-specific differences (e.g. ARIMA's d=0 vs d=1 distinction).

**3. Random Forest cannot extrapolate beyond its training range — and this is a structural limitation, not a tuning problem.**
This was the most important model-comparison finding of the project. For products and warehouses with stable or smoothly trending demand, Random Forest performed competitively or even outperformed ARIMA. But for series with a structural break — most clearly Product_1341, where 2016 demand collapsed below any value seen in 2013–2015 training data — RF's accuracy collapsed (140.6% MAPE vs. ARIMA's 51.3%). Hyperparameter tuning (Task 9) confirmed this: no combination of `n_estimators`, `max_depth`, or `min_samples_leaf` meaningfully closed this gap, because the issue is inherent to how decision trees make predictions (by averaging training examples, not extrapolating a trend line the way ARIMA can).

**4. Hyperparameter tuning's value depends heavily on the underlying data, not just the model type.**
For the warehouse-level Random Forest model, every tuned configuration performed equal to or worse than the default — the default was already close to its practical ceiling for that feature set. For the product-level model, tuning produced a real (if modest) improvement, likely because the smaller, more volatile product-level dataset benefited from added regularization (`max_depth` limiting, higher `min_samples_leaf`) that reduced overfitting. The broader lesson: a tuning experiment that "fails" to improve one model variant is still a useful, reportable finding — it tells you where the model's error ceiling actually comes from (data/features vs. configuration).

**5. Residual analysis reveals systematic bias that a single MAPE number hides.**
Two models could share a similar overall MAPE while having very different error *patterns*. Whse_S's residuals were almost entirely negative across all of 2016 (a consistent over-forecasting bias), while Whse_C's and Whse_A's residuals were balanced in both directions (closer to random noise). This distinction matters operationally: a systematic bias can be corrected with a simple adjustment factor, while balanced noise cannot — meaning the same MAPE score can imply two very different next steps depending on the residual pattern behind it.

**6. Different forecasting libraries fitting the same model order can still produce different forecasts.**
An early discovery was that `pmdarima` and `statsmodels`, even when fit with identical (p,d,q) orders, produced different forecasts due to differences in their internal optimization routines. The practical lesson: forecasts must be generated and saved directly from the library used for fitting, not refit separately in another library expecting identical results — this is a subtle reproducibility trap worth documenting for future model deployment work.

### Recommendations for Future Work

1. **Expand the feature set rather than further tuning Random Forest.** Since the warehouse-level model's hyperparameter tuning showed no improvement, the more promising direction is incorporating features the model currently lacks — e.g. promotions, pricing changes, or other external demand drivers.
2. **Add structural-break detection.** Since Random Forest's main failure mode is an inability to extrapolate beyond training range, a monitoring step that flags when recent actuals fall outside the model's training distribution would catch a future Whse_S/Product_1341-style shift early, before forecast quality silently degrades.
3. **Compare against the team's other models (SARIMA, XGBoost, Prophet).** This project compared ARIMA and Random Forest in depth; extending the same evaluation framework (test-set MAE/MAPE/RMSE, residual analysis) to the team's other models would clarify whether seasonal-aware or boosting-based approaches close the gap on the hardest series.
4. **Use a hybrid deployment by default for problems with this structure.** No single model dominated across all warehouses and products — the practical recommendation is to select a model per series based on its demand characteristics (stationary vs. trending vs. structurally shifting), rather than committing to one algorithm for the entire dataset.
